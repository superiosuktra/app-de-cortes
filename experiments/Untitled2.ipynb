### Aplicação de Correção com Expressões Regulares (Regex)
Usando Regex, podemos encontrar de forma flexível o bloco de código que executa o `yt-dlp` e aplicar o tratamento de erros sem que quebras de linha específicas impeçam a substituição.

```python
import os
import re

pasta_projeto = '/content/ai-viral-shorts-cutter-8-de-out-de-2026-15-49-48_extraido'
server_path = os.path.join(pasta_projeto, 'server.ts')

with open(server_path, 'r', encoding='utf-8') as f:
    conteudo = f.read()

# Procura o trecho de download do yt-dlp usando uma expressão regular flexível
padrao = r"(\s*// 2\. Try yt-dlp slice.*?)(if \(fs\.existsSync\(ytdlpPath\)\) \{[\s\S]*?)(await execPromise\(dlCmd[^;]*;[\s\S]*?\}\s*\}\s*\})"

match = re.search(padrao, conteudo)
if match:
    # Estruturamos o novo bloco com try/catch e logs aprimorados
    novo_bloco = """    // 2. Try yt-dlp slice if YouTube URL is provided and yt-dlp binary exists
    if (
      videoUrl &&
      (videoUrl.includes('youtube.com') || videoUrl.includes('youtu.be'))
    ) {
      if (fs.existsSync(ytdlpPath)) {
        try {
          console.log(`[YouTube Download] Iniciando download via yt-dlp para a URL: ${videoUrl}`);
          const rawSlice = path.join(uploadsDir, `raw_slice_${Date.now()}.mp4`);
          const dlCmd = `"${ytdlpPath}" --js-runtimes node:/usr/local/bin/node --no-playlist -f "18/bestvideo[height<=720][ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best" --download-sections "*${startTime}-${endTime}" "${videoUrl}" -o "${rawSlice}" --socket-timeout 15`;

          await execPromise(dlCmd, { timeout: 35000 });

          if (fs.existsSync(rawSlice)) {
            videoPath = rawSlice;
            console.log('[YouTube Download] Download concluído com sucesso. Aplicando corte vertical 9:16 via FFmpeg...');
            const cutCmd = `ffmpeg -y -i "${rawSlice}" -vf "crop=ih*(9/16):ih" -c:a copy "${outMp4Path}"`;
            await execPromise(cutCmd);
            hasCutGenerated = true;
          } else {
            throw new Error('O arquivo de slice temporário não foi criado pelo yt-dlp.');
          }
        } catch (dlErr: any) {
          console.error('[YouTube Download Erro] Falha no download ou processamento do yt-dlp:', dlErr.message || dlErr);
          // Fallback seguro
          console.warn('yt-dlp slice notice (falling back to animated 9:16 template devido à falha anterior)');
        }
      } else {
        console.error(`[YouTube Download Erro] Binário do yt-dlp não foi encontrado no caminho esperado: ${ytdlpPath}`);
      }
    }"""

    conteudo_atualizado = re.sub(padrao, novo_bloco, conteudo, count=1)
    with open(server_path, 'w', encoding='utf-8') as f:
        f.write(conteudo_atualizado)
    print("Sucesso! O tratamento de erros foi injetado com precisão via Regex no arquivo server.ts.")
else:
    print("Não foi possível mapear o bloco padrão com Regex. Vamos imprimir as linhas próximas a 'ytdlpPath' para ajuste manual.")
    linhas = conteudo.splitlines()
    for idx, linha in enumerate(linhas):
      if 'ytdlpPath' in linha:
        print(f"Linha {idx+1}: {linha}")
```

### Guia para Enviar as Alterações ao GitHub

Como o ambiente do Colab é temporário, para enviar as alterações que fizemos de volta para o seu repositório no GitHub, você precisará configurar suas credenciais do Git (Nome, Email e o seu Personal Access Token do GitHub).

```python
#@title Configurar Credenciais e Enviar ao GitHub (Opcional)

# Substitua com os seus dados do GitHub
GITHUB_USER = "seu-usuario"
GITHUB_REPO = "seu-repositorio"
GITHUB_TOKEN = "seu_personal_access_token_aqui"
EMAIL = "seu-email@exemplo.com"
NOME = "Seu Nome"

import os

# Caminho da pasta extraída do projeto
pasta_projeto = '/content/ai-viral-shorts-cutter-8-de-out-de-2026-15-49-48_extraido'
os.chdir(pasta_projeto)

print("--- Configurando Git local ---")
os.system(f'git config --global user.email "{EMAIL}"')
os.system(f'git config --global user.name "{NOME}"')

# Inicializa ou atualiza o repositório git local
if not os.path.exists('.git'):
    os.system('git init')
    os.system('git branch -M main')

# Adiciona os arquivos modificados
os.system('git add .')
os.system('git commit -m "Fix: Adiciona mocks e corrige erro de build do recharts (react-is)"')

# Configura a URL remota de forma segura usando o Token
remote_url = f"https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{GITHUB_REPO}.git"
os.system(f'git remote remove origin 2>/dev/null')
os.system(f'git remote add origin {remote_url}')

print("\nPronto! Para enviar as alterações ao GitHub, execute em uma nova célula:")
print("!git push -u origin main")
```
